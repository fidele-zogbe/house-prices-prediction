Take all the numerical columns and find their correlation

```python
num = df_train.select_dtypes(include='number')
num
```

```python
corr = num.corr()
corr
```

```python
corr_ = corr['SalePrice'].drop("SalePrice").abs().sort_values(ascending=False)
corr_
```

```python
def get_targetcorr(value, corre, target):
    target_corr = corre[target].drop(target)
    high_corr = target_corr[target_corr.abs() > value].sort_values(key = abs, ascending = False)

    return target_corr, high_corr
```

```python
get_targetcorr(0.7, corr, "SalePrice")[1]
```

```python
get_targetcorr(0.7, corr, "SalePrice")[0]
```

```python
import numpy as np

features = num.drop(columns= ["Id", "SalePrice"], errors="ignore")
target = num.corr()['SalePrice'].abs()

corr_features = features.corr().abs()
upper = corr_features.where(np.triu(np.ones(corr_features.shape), k=1).astype(bool))
pairs = upper.stack()
pairs = pairs[pairs>0.7].sort_values(ascending=False)

to_drop = set()
pairs
```

```python
for (a, b), r in pairs.items():
    if a in to_drop or b in to_drop:
        continue
    to_drop.add(a if target[a] < target[b] else b)

to_drop
```

```python
df_train = df_train.drop(columns=to_drop)
df_test = df_test.drop(columns=to_drop)
df_train
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X = df_train.drop(columns=['SalePrice', 'Id'], errors='ignore')
y = df_train.SalePrice
```

```python
X_train, X_val, y_train, y_val = train_test_split(X, y, random_state=42, test_size=0.2)
cat_cols = list(X_train.select_dtypes(include='object').columns)
cat_cols
```

```python
ct = ColumnTransformer(
    transformers=[('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)], 
    remainder="passthrough")

X_train_enc = ct.fit_transform(X_train)
X_val_enc = ct.transform(X_val)
```

## keep in mind that transform is only used on test.

```python
X_train_enc.shape
```

```python
X_val_enc.shape
```

```python
rf_model = RandomForestRegressor()
rf_model.fit(X_train_enc, y_train)
pred = rf_model.predict(X_val_enc)
```

```python
mean_absolute_error(pred, y_val)
```

```python
r2_score(y_val, pred)
```

```python
np.sqrt(mean_squared_error(y_val, pred))
```

```python
pred
```

when I reduce the correlation to 0.7 (it means that every variable that correlates with others
with 0.7 correlation rate should be dropped), the r2_score increased.

When 0.69, 0.8904

when 0.68, 0.88055

When decreased to 0.6, r2_score decreased to ~0.87

But when I reduced it to 0.5, the r2_score decreased to from around 0.9 to 0.84

```python
print(r2_score(y_val, rf_model.predict(X_val_enc)))
print(r2_score(y_train, rf_model.predict(X_train_enc)))
```
